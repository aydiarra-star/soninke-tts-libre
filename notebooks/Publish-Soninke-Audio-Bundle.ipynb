# Publier les audios du lexique Soninké TTS Libre

Ce notebook utilise l’archive existante `soninke_audio_export.zip` : il la cherche d’abord dans Google Drive, puis permet de sélectionner le ZIP depuis l’appareil si elle n’y est pas. Il vérifie le manifeste et les WAV, prépare le lecteur audio, publie une branche avec les fichiers réels et ouvre une PR vers `main`. **Aucune fusion automatique.**

Prérequis : archive dans `MyDrive/SONINKE_TTS_LIBRE/soninke_audio_export.zip`. Il faut un jeton GitHub temporaire limité au dépôt `aydiarra-star/xamxam-xara`, avec **Contents: Read and write** et **Pull requests: Read and write**. La saisie du jeton sera masquée. Aucun audio n’est régénéré et aucune validation linguistique n’est redemandée.

```python
from urllib.request import urlopen
url = 'https://raw.githubusercontent.com/aydiarra-star/xamxam-xara/automation/audio-bundle-publisher/scripts/publish_audio_bundle_colab.py'
with urlopen(url, timeout=30) as response:
    source = response.read().decode('utf-8')
print('Automatisation récupérée. Le script va vérifier le ZIP avant toute publication.')
exec(compile(source, 'publish_audio_bundle_colab.py', 'exec'), {'__name__': '__main__'})
```
